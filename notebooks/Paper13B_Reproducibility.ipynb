# Paper 13B reproducibility notebook

This notebook reproduces the checkpoint-replication, stability, multilingual, and mitigation summaries from the archived derived tables.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
ROOT = Path('..').resolve()
TABLES = ROOT / 'tables'
FIGURES = ROOT / 'figures'
print(ROOT)
```

## Cross-checkpoint construct collapse

```python
gfac = pd.read_csv(TABLES / 't29_general_factor_share.csv')
halo = pd.read_csv(TABLES / 't30_llm_family_halo.csv')
agreement = pd.read_csv(TABLES / 't31_qwen_vs_mistral_agreement.csv')
display(gfac); display(halo); display(agreement)
```

```python
d = halo[halo['corpus'].eq('all')].copy()
ax = d.plot.bar(x='method', y='mean_halo_r', legend=False, color='#7f3c8d', figsize=(8,4))
ax.set_ylim(0,1); ax.set_ylabel('Mean heterotrait correlation'); ax.set_title('Method halo by measurement arm')
plt.tight_layout(); plt.show()
```

## Stability, prompts, and language

```python
stability = pd.read_csv(TABLES / 't24_llm_run_stability.csv')
prompt = pd.read_csv(TABLES / 't25_llm_prompt_sensitivity.csv')
language = pd.read_csv(TABLES / 't27_multilingual_invariance.csv')
display(stability); display(prompt); display(language)
```

## Elicitation interventions

```python
mitigation = pd.read_csv(TABLES / 't33_mitigation_profile.csv')
coverage = pd.read_csv(TABLES / 't34_mitigation_coverage.csv')
quotes = pd.read_csv(TABLES / 't35_evidence_quote_audit.csv')
display(mitigation); display(coverage); display(quotes)
```

## Published manuscript figures

```python
figure_names = ['f08_llm_family_halo.png','f08_llm_stability.png','f09_perturbation_audit.png','f11_mitigation.png']
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
for ax, name in zip(axes.flat, figure_names):
    ax.imshow(plt.imread(FIGURES / name)); ax.set_title(name); ax.axis('off')
plt.tight_layout(); plt.show()
```
